```python
import pandas as pd
import numpy as np
import numpy as np

# Load dataset
file_path = 'Used_Car_Price_Prediction.csv'
df = pd.read_csv(file_path)

# Display total rows and columns
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n")
```

```python

# Separate columns into lists
numerical_cols = []
categorical_cols = []

for column in df.columns:
    if df[column].dtype in ['int64', 'float64']:
        numerical_cols.append(column)
    else:
        categorical_cols.append(column)

# Output results
print("NUMERICAL COLUMNS:")
print(numerical_cols)

print("\nCATEGORICAL COLUMNS:")
print(categorical_cols)
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns

# 3. EDA on Numerical Columns
print("=" * 40)
print("NUMERICAL COLUMNS EDA")
print("=" * 40)

for col in num_cols:
    print(f"Column: {col}")
    print(f"  Mean:   {df[col].mean():.2f}")
    print(f"  Median: {df[col].median():.2f}")
    print(f"  Min:    {df[col].min()}")
    print(f"  Max:    {df[col].max()}")
    print(f"  Std:    {df[col].std():.2f}\n")

# 4. EDA on Categorical Columns
print("=" * 40)
print("CATEGORICAL COLUMNS EDA")
print("=" * 40)

for col in cat_cols:
    most_frequent = df[col].mode().iloc[0]
    frequency = df[col].value_counts().iloc[0]
    
    print(f"Column: {col}")
    print(f"  Unique Values: {df[col].nunique()}")
    print(f"  Most Frequent: {most_frequent} ({frequency} times)\n")
```

```python
cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns

# 3. EDA on Categorical Columns
print("=" * 45)
print("CATEGORICAL & BOOLEAN COLUMNS EDA")
print("=" * 45)

for col in cat_cols:
    unique_count = df[col].nunique()
    missing_count = df[col].isnull().sum()
    top_value = df[col].mode().iloc[0]
    top_frequency = df[col].value_counts().iloc[0]
    
    print(f"Column Name: {col}")
    print(f"  - Unique Values : {unique_count}")
    print(f"  - Missing Values: {missing_count}")
    print(f"  - Top Category  : {top_value} ({top_frequency} times)\n")
```

```python
# View first 5 rows
print("--- First 5 Rows ---")
print(df.head())

# View basic info on data types
print("\n--- Data Types & Non-Null Counts ---")
print(df.info())
```

```python
# Check duplicate rows
duplicates = df.duplicated().sum()
print(f"Duplicate Rows Count: {duplicates}\n")

# Calculate missing values count and percentage
missing_data = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage (%)': np.round((df.isnull().sum() / len(df)) * 100, 2)
})

# Filter to show only columns with missing values
missing_data = missing_data[missing_data['Missing_Count'] > 0].sort_values(
    by='Missing_Count', ascending=False
)

print("--- Missing Values Summary ---")
print(missing_data)
```

```python
# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Standard summary statistics
num_summary = df[num_cols].describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']]

# Calculate skewness for numerical variables
num_summary['skewness'] = df[num_cols].skew()

print("--- Numerical Features Summary Statistics ---")
print(np.round(num_summary, 2))
```

```python
# Correlation matrix relative to sale_price
if 'sale_price' in df.columns:
    corr = df[num_cols].corr()['sale_price'].sort_values(ascending=False)
    print("--- Pearson Correlation with Sale Price ---")
    print(np.round(corr, 3))
```

```python
# Select categorical and boolean columns
cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns

print("--- Categorical & Boolean Summary ---")
cat_summary = df[cat_cols].describe().T
print(cat_summary)

# Display top value counts for key categorical features
print("\n--- Top Category Frequencies ---")
for col in cat_cols:
    n_unique = df[col].nunique(dropna=False)
    print(f"\nFeature: {col} (Unique Values: {n_unique})")
    print(df[col].value_counts(dropna=False).head(5))
```

```python
print("--- Data Quality Checks & Zero-Value Counts ---")

# Check for 0 values in price/financing columns
check_cols = ['sale_price', 'broker_quote', 'emi_starts_from', 'booking_down_pymnt', 'kms_run']
for col in check_cols:
    if col in df.columns:
        zero_count = (df[col] == 0).sum()
        print(f"Zero values in '{col}': {zero_count}")

# IQR-based outlier count for numerical features
print("\n--- Outlier Counts (1.5 * IQR Rule) ---")
for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    print(f"Feature '{col}': {outliers} outliers")
```
